```python
import pandas as pd
import numpy as np

# Load dataset
file_path = r"C:\Users\Lenovo\Desktop\Mahak Project\Marketing Campaign Business Performance Analytics\data\raw\ppc_campaign_performance_data.xlsx"

df = pd.read_excel(file_path)

# Convert Date
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Recalculate core marketing KPIs
df["Calculated_CTR"] = df["Clicks"] / df["Impressions"]
df["Calculated_CPC"] = df["Spend"] / df["Clicks"]
df["Calculated_Conversion_Rate"] = df["Conversions"] / df["Clicks"]
df["Calculated_CPA"] = df["Spend"] / df["Conversions"]
df["Calculated_ROAS"] = df["Revenue"] / df["Spend"]

# Compare supplied vs calculated metrics
df["CTR_Difference"] = df["CTR"] - df["Calculated_CTR"]
df["CPC_Difference"] = df["CPC"] - df["Calculated_CPC"]
df["Conversion_Rate_Difference"] = (
    df["Conversion_Rate"] - df["Calculated_Conversion_Rate"]
)
df["CPA_Difference"] = df["CPA"] - df["Calculated_CPA"]
df["ROAS_Difference"] = df["ROAS"] - df["Calculated_ROAS"]

# Validation summary
print("Dataset shape:", df.shape)
print("Duplicate rows:", df.duplicated().sum())
print("\nDate range:")
print(df["Date"].min(), "to", df["Date"].max())

print("\nNumeric validation:")
display(
    df[
        [
            "CTR_Difference",
            "CPC_Difference",
            "Conversion_Rate_Difference",
            "CPA_Difference",
            "ROAS_Difference"
        ]
    ].describe()
)
```

```python
# Show rows where supplied metrics differ noticeably from recalculated values

validation = df[
    [
        "Campaign_ID",
        "CTR",
        "Calculated_CTR",
        "CPC",
        "Calculated_CPC",
        "Conversion_Rate",
        "Calculated_Conversion_Rate",
        "CPA",
        "Calculated_CPA",
        "ROAS",
        "Calculated_ROAS"
    ]
].copy()

validation["CPA_Abs_Difference"] = (
    validation["CPA"] - validation["Calculated_CPA"]
).abs()

validation["ROAS_Abs_Difference"] = (
    validation["ROAS"] - validation["Calculated_ROAS"]
).abs()

display(
    validation.sort_values(
        "CPA_Abs_Difference",
        ascending=False
    ).head(10)
)
```
